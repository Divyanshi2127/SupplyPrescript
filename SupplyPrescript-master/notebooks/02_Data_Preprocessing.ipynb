# Notebook 2: Data Cleaning & Feature Engineering

## Objective

The goal of this notebook is to prepare the supply chain dataset for machine learning and optimization.

Tasks covered:
- Remove unnecessary columns
- Handle missing values
- Remove duplicates
- Convert data types
- Engineer new features
- Encode categorical variables
- Save the cleaned dataset

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
```

```python


df = pd.read_csv(
    "C:/Users/USER/OneDrive/Desktop/VSCode/c++/SupplyPrescript/data/Raw/SupplyChain/DataCoSupplyChainDataset.csv",
    encoding="latin1"
)
```

```python
df.head()
```

```python
data=df.copy()
```

```python
print("Duplicate rows:", data.duplicated().sum())

data = data.drop_duplicates()
```

```python
data.columns
```

```python
columns_to_drop = [

    "Customer Email",
    "Customer Password",
    "Customer Fname",
    "Customer Lname",
    "Customer Street",
    "Customer Zipcode",

    "Product Description",
    "Product Image",

    "Latitude",
    "Longitude",

    "Order Zipcode",

    "Order Customer Id",
    "Customer Id",

    "Category Id",
    "Department Id",

    "Product Card Id",
    "Product Category Id",

    "Order Item Cardprod Id",
    "Order Item Id",

    "Order Id"

]

data.drop(columns=columns_to_drop, inplace=True)
```

```python
data.columns
```

```python
data["order date (DateOrders)"] = pd.to_datetime(
    data["order date (DateOrders)"]
)
data["shipping date (DateOrders)"] = pd.to_datetime(
    data["shipping date (DateOrders)"]
)
```

```python
data["Lead_Time"] = (
    data["shipping date (DateOrders)"] -
    data["order date (DateOrders)"]
).dt.days
```

```python
data["Order_Month"] = data["order date (DateOrders)"].dt.month
```

```python
data["Order_Year"] = data["order date (DateOrders)"].dt.year
```

```python
data["Order_Weekday"] = data["order date (DateOrders)"].dt.day_name()
```

```python
data["Profit_Margin"] = (
    data["Benefit per order"] /
    data["Sales"]
)
```

```python
data["High_Value_Order"] = (
    data["Sales"] >
    data["Sales"].median()
).astype(int)
```

```python
data.drop(
    columns=[
        "order date (DateOrders)",
        "shipping date (DateOrders)"
    ],
    inplace=True
)
```

```python
numeric = data.select_dtypes(include=np.number).columns

for col in numeric:
    data[col].fillna(data[col].median(), inplace=True)
```

```python
categorical = data.select_dtypes(include="object").columns

for col in categorical:
    data[col].fillna(data[col].mode()[0], inplace=True)
```
