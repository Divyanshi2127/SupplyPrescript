# Notebook 4: Feature Selection & Model Preparation

## Objective

This notebook prepares the dataset for predictive modeling by:

- Selecting the target variable
- Removing data leakage
- Selecting useful features
- Encoding categorical variables
- Splitting data into training and testing sets
- Scaling numerical features (if needed)
- Saving the prepared datasets for model training

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("../data/processed/clean_supply_chain.csv")

data.head()
```

```python
data.info()
```

```python
data.shape
```

```python

target = "Late_delivery_risk"
```

```python
data[target].value_counts()
```

```python
leakage_columns = [

    "Late_delivery_risk","Delivery Status","Days for shipping (real)","Order Status"                 
]
```

```python
data = data.drop(columns=leakage_columns)
```

```python
id_columns = [

    "Category Id",
    "Department Id",
    "Order Id",
    "Order Item Id",
    "Order Item Cardprod Id",
    "Product Card Id",
    "Product Category Id",
    "Customer Id",
    "Order Customer Id"
]
```

```python
id_columns = [col for col in id_columns if col in data.columns]

data = data.drop(columns=id_columns)
```

```python
data.columns
```

```python
categorical = data.select_dtypes(include="object").columns

categorical
```

```python
encoder = LabelEncoder()

for col in categorical:
    data[col] = encoder.fit_transform(data[col])
```

```python
full_data = pd.read_csv("../data/processed/clean_supply_chain.csv")

y = full_data["Late_delivery_risk"]
```

```python
X = data
```

```python
print(X.shape)
print(y.shape)
```

```python
y.value_counts()
```

```python
y.value_counts(normalize=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)
```

```python
print("Training Features :", X_train.shape)

print("Testing Features :", X_test.shape)

print("Training Labels :", y_train.shape)

print("Testing Labels :", y_test.shape)
```

```python
X_train.to_csv("../data/processed/X_train.csv", index=False)

X_test.to_csv("../data/processed/X_test.csv", index=False)

y_train.to_csv("../data/processed/y_train.csv", index=False)

y_test.to_csv("../data/processed/y_test.csv", index=False)
```

```python
feature_list = pd.DataFrame({

    "Feature": X.columns

})

feature_list
```

```python
feature_list.to_csv(

    "../data/processed/features.csv",

    index=False

)
```
